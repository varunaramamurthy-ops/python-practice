```python
amt_of_butter_user_has = float(input("Enter the amount of butter you have in grams: "))
butter_needed = 115  # grams
butter_short_by = butter_needed - amt_of_butter_user_has
extra_butter = amt_of_butter_user_has - butter_needed

if amt_of_butter_user_has < butter_needed:
    print(f"Yourbutter is short by {butter_short_by} grams.")

elif amt_of_butter_user_has == butter_needed:
    print("You have the perfect amount of butter.")

else:
    print(f"You have {extra_butter} grams of extra butter.")
```
