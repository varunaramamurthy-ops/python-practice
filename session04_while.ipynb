```python
butter_available = float(input("Enter the amount of butter you have in grams: "))
butter_needed = 115  # grams
while butter_available < 0:
    print("Butter can't be negative. Try again.")
    butter_available = float(input("Enter the amount of butter you have in grams: "))

if butter_available < butter_needed: 
    butter_short_by = butter_needed - butter_available
    print(f"Your butter is short by {butter_short_by} grams.")

elif butter_available == butter_needed:
    print("You have the perfect amount of butter.")

else:
    extra_butter = butter_available - butter_needed
    print(f"You have {extra_butter} grams of extra butter.")
```
